## Redo Seeding

```python
# Create path

path_seed = r'/Users/tyrasmussen/Desktop/FVTC/Spring 2026/Data Analytics Capstone/MM/MM Data/2002 - 2025/Data/'


```

```python
# Import seed data to add to dataframe

df_seed = pd.read_csv(os.path.join(path_seed, "REF _ Post-Season Tournament Teams.csv"))
```

```python
df_seed.head()
```

```python
# Keep only rows where Post-Season_Tournament = 'March Madness' and seed is not null
df_seed = df_seed[(df_seed['Post-Season Tournament'] == 'March Madness') & (df_seed['Seed'].notnull())]
```

```python
# Keep only columns for Season, Team Name, and Seed
df_seed = df_seed[['Season', 'Team Name', 'Seed']]
```

```python
# Check new dataframe
df_seed.head()
```

```python
# Show number of values by Seed and Season
seed_counts_seed_data = df_seed.groupby(['Season', 'Seed']).size().unstack(fill_value=0)
print(seed_counts_seed_data)
```

```python
# Join df_seed with df on season and team_name to add seed information to main dataframe
df = df.merge(df_seed, left_on=['season', 'team_name'], right_on=['Season', 'Team Name'], how='left')
```

```python
# Remove new columns 'Season' and 'Team Name' from df and old 'seed' column from df to avoid confusion
df = df.drop(columns=['Season', 'Team Name', 'seed'])
```

```python
df.head()
```

```python
# Count values of seed column by season to confirm changes
seed_counts = df.groupby(['season', 'Seed']).size()
print(seed_counts)
```

```python
# Rename new 'Seed' column to 'seed' for consistency
df = df.rename(columns={'Seed': 'seed'})
```

```python
# Show columns
print(df.columns)
```

```python
# Rename 'adjusted_temo' column to 'adjusted_tempo' for consistency
df = df.rename(columns={'adjusted_temo': 'adjusted_tempo'})
```
